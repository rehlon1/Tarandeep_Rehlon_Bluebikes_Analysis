```python
weather_summary_df = weather_summary_df.merge(
weather_mapping_df[["station_id", "district"]],
left_on="STATION",
right_on="station_id",
how="left"
).drop(columns="station_id")
```

```python
bluebikes_df = pd.read_csv('Final BlueBikes Data.csv')
```

```python
bluebikes_df
```

```python
bluebikes_df.shape
```

```python
(bluebikes_df.isnull().sum())
```

```python
bluebikes_df.dtypes
```

```python
bluebikes_df["start_time"] = pd.to_datetime(bluebikes_df["start_time"],format="mixed")
```

```python
bluebikes_df.dtypes
```

```python
bluebikes_df["DATE"] = pd.to_datetime(bluebikes_df["start_time"]).dt.normalize()
```

```python
grouped_bluebikes_df = bluebikes_df.groupby(["DATE", "start_jrny_district"]).size().reset_index(name="Total Journeys")
```

```python
grouped_bluebikes_df
```

```python
final_weather_df = grouped_bluebikes_df.merge(
weather_summary_df[["DATE", "district", "PRCP", "SNOW", "SNWD", "TMAX", "TMIN"]],
left_on= ["DATE", "start_jrny_district"], 
right_on=["DATE", "district"],
how="left"
)
```

```python
final_weather_df
```

```python
final_weather_df.isnull().sum()
```

```python
final_weather_df = final_weather_df.rename(columns={
    "PRCP": "Rainfall (mm)",
    "SNOW": "Snowfall (mm)",
    "SNWD": "Snow on ground (mm)",
    "TMAX": "Max Temp (°C)",
    "TMIN": "Min Temp (°C)"
})
```
