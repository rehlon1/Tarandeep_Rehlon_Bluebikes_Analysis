```python
import pandas as pd
```

```python
bluebikes_df = pd.read_csv('Final BlueBikes Data.csv')
```

```python
bluebikes_df
```

```python
start_journey_df = bluebikes_df.groupby(['start_jrny_name'])['bike_id'].count()
```

```python
start_journey_df = start_journey_df.reset_index(name="Total Starts")
```

```python
start_journey_df = start_journey_df.rename(columns={'start_jrny_name': 'Station name'})
```

```python
start_journey_df
```

```python
end_journey_df = bluebikes_df.groupby(['end_jrny_name'])['bike_id'].count()
```

```python
end_journey_df = end_journey_df.reset_index(name="Total Ends")
```

```python
end_journey_df = end_journey_df.rename(columns={'end_jrny_name': 'Station name'})
```

```python
end_journey_df
```

```python
combined_journey_df = start_journey_df.merge(
end_journey_df[["Station name", "Total Ends"]],
on= ["Station name"], 
how="outer")
```

```python
combined_journey_df
```

```python
combined_journey_df.to_csv('Station Start_End Data.csv')
```
